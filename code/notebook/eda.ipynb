```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../../dataset/df_features.csv')
df.info()
```

```python
df.head()
```

```python
df['room'].value_counts().plot(kind='bar')
```

```python
sns.boxplot(x='hour', y='room', data=df)
```

```python
rssi_cols = [c for c in df.columns if '_min' in c or '_max' in c or '_count' in c]
df[rssi_cols].describe()
```

```python

sns.heatmap(df[rssi_cols].corr(), cmap='coolwarm')
```

```python
# Pisahkan fitur RSSI
rssi_cols = [c for c in df.columns if '_min' in c or '_max' in c or '_count' in c]

# 1. Statistik ringkas per room
room_stats = df.groupby('room')[rssi_cols].agg(['mean', 'std', 'min', 'max'])
room_stats
```

```python
# 2. Visualisasi distribusi RSSI min/max untuk beberapa beacon penting
beacons_sample = rssi_cols[:6]  # ambil 6 beacon pertama sebagai contoh
for beacon in beacons_sample:
    plt.figure(figsize=(8,4))
    sns.boxplot(x='room', y=beacon, data=df)
    plt.title(f'Distribusi {beacon} per Room')
    plt.xticks(rotation=45)
    plt.show()
```

```python
# 3. Korelasi beacon per room (contoh room pertama)
sample_room = df['room'].unique()[0]
plt.figure(figsize=(10,8))
sns.heatmap(df[df['room']==sample_room][rssi_cols].corr(), cmap='coolwarm')
plt.title(f'Korelasi RSSI Beacon di Room {sample_room}')
plt.show()
```

```python
# 4. Analisis waktu (hour) per room
plt.figure(figsize=(10,4))
sns.countplot(x='hour', hue='room', data=df)
plt.title('Distribusi waktu (hour) per room')
plt.show()
```

```python
# 5. Beacon terdeteksi (count > 0) per room
count_cols = [c for c in df.columns if '_count' in c]
room_beacon_count = df.groupby('room')[count_cols].mean()
plt.figure(figsize=(12,6))
sns.heatmap(room_beacon_count, annot=True, fmt=".2f", cmap='YlGnBu')
plt.title('Rata-rata deteksi beacon per room')
plt.show()
```
